```python
import base64, io, os, subprocess, sys, zipfile
PAYLOAD = ''
ROOT = "/tmp/gpf"
zipfile.ZipFile(io.BytesIO(base64.b64decode(PAYLOAD))).extractall(ROOT)
subprocess.run(["nvidia-smi", "-L"])
procs = []
for gpu, part in enumerate([['cdp101', 'cdp203', 'rdp103', 'rdp201', 'rcdp101', 'rcdp202'], ['cdp103', 'cdp207', 'rdp107', 'rdp210', 'rcdp104', 'rcdp207']]):
    env = dict(os.environ, CUDA_VISIBLE_DEVICES=str(gpu))
    log = open(f"/kaggle/working/gpu{gpu}.log", "w")
    procs.append(subprocess.Popen([sys.executable, os.path.join(ROOT, "full_run.py"), ROOT, "src_python_gpu_SA_RCRS_GRASP",
                                   os.path.join(ROOT, "dataset"), "/kaggle/working/results", "B"] + part,
                                  env=env, stdout=log, stderr=subprocess.STDOUT))
for p in procs:
    p.wait()
print(open("/kaggle/working/gpu0.log").read()[-3000:]); print(open("/kaggle/working/gpu1.log").read()[-3000:])
```
