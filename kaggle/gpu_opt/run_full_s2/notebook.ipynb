```python
import base64, io, os, subprocess, sys, zipfile
PAYLOAD = ''
ROOT = "/tmp/gpf"
zipfile.ZipFile(io.BytesIO(base64.b64decode(PAYLOAD))).extractall(ROOT)
subprocess.run(["nvidia-smi", "-L"])
import json
logs = [open(f"/kaggle/working/gpu{g}.log", "w") for g in (0, 1)]
for spec in [{'name': 'sa', 'cfg': {'init': 'sa'}}, {'name': 'grasp_nosa', 'cfg': {'sa_warmup_iterations': 0}}, {'name': 'grasp_light', 'cfg': {'sa_warmup_iterations': 10}, 'i2': {'SA_T0': 0.01, 'SA_COOLING': 0.5, 'SA_TMIN': 0.0005, 'SA_RELATIVE': True}}, {'name': 'mixed', 'cfg': {'init': 'mixed'}}, {'name': 'mixed_nosa', 'cfg': {'init': 'mixed', 'sa_warmup_iterations': 0}}, {'name': 'mixed_light', 'cfg': {'init': 'mixed', 'sa_warmup_iterations': 10}, 'i2': {'SA_T0': 0.01, 'SA_COOLING': 0.5, 'SA_TMIN': 0.0005, 'SA_RELATIVE': True}}]:
    procs = []
    for gpu, part in enumerate([['cdp101', 'cdp203', 'rdp103', 'rdp201', 'rcdp101', 'rcdp202'], ['cdp103', 'cdp207', 'rdp107', 'rdp210', 'rcdp104', 'rcdp207']]):
        env = dict(os.environ, CUDA_VISIBLE_DEVICES=str(gpu), CFG=json.dumps(spec.get("cfg", {})),
                   I2=json.dumps(spec.get("i2", {})))
        procs.append(subprocess.Popen([sys.executable, os.path.join(ROOT, "full_run.py"), ROOT, "src_python_gpu_SA_RCRS_GRASP",
                                       os.path.join(ROOT, "dataset"), "/kaggle/working/results", spec["name"]] + part,
                                      env=env, stdout=logs[gpu], stderr=subprocess.STDOUT))
    for p in procs:
        p.wait()
for f in logs:
    f.close()
print(open("/kaggle/working/gpu0.log").read()[-3000:]); print(open("/kaggle/working/gpu1.log").read()[-3000:])
```
